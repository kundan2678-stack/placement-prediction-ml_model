```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("college_student_placement_dataset.csv")
```

```python
data.head()
```

```python
data = data.drop(["College_ID"],axis=1)
```

```python
data.head()
```

```python
data = data.replace(to_replace= ["No","Yes"],value=[0.0,1.0])
```

```python
y = np.array(data["CGPA"],dtype=np.float32)
x = np.array(data.drop(["CGPA"],axis=1),dtype=np.float32)
```

```python
x_transpose = np.transpose(x)
x_transpose_x = np.matmul(x_transpose,x)
x_transpose_x_inv = np.linalg.inv(x_transpose_x)
x_transpose_x_T = np.matmul(x_transpose_x_inv,x_transpose)
w_hat = np.matmul(x_transpose_x_T,y)
```
